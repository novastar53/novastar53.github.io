# How Gradient Variance Shapes Policy Optimization

*Why unbiased policy-gradient estimates can lead to dramatically different learning outcomes.*

In this blogpost, we investigate how reward baselines affect the learning dynamics of policy-gradient algorithms applied to multi-armed bandit problems.

Multi-armed bandits are a class of single-state decision problems in which an agent selects one of a fixed set of actions, each producing a stochastic reward. Unlike value-based methods, which estimate the expected reward associated with each action, policy-gradient methods directly optimize a parameterized policy to maximize expected reward.

The optimization objective is:

$$
J(\theta) = \mathbb{E}_{\pi_\theta}[R]
$$

where $\theta$ represents the policy parameters.

We can optimize this objective using gradient ascent:

$$
\theta_{t+1} = \theta_t + \alpha \nabla_\theta J(\theta_t)
$$

where $\alpha$ is the learning rate.

However, the reward distributions are unknown, so we cannot directly evaluate the gradient of the expected reward. Fortunately, the log-derivative trick allows us to express the gradient in a form that can be estimated from sampled actions and rewards.

For a multi-armed bandit:

$$
J(\theta) = \sum_a \pi_\theta(a)q_*(a)
$$

where $q_*(a)$ is the true expected reward associated with action $a$.

Differentiating with respect to $\theta$:

$$
\begin{aligned}
\nabla_\theta J(\theta)
&= \sum_a q_*(a)\nabla_\theta \pi_\theta(a) \\
&= \sum_a \pi_\theta(a)q_*(a)\nabla_\theta \log \pi_\theta(a) \\
&= \mathbb{E}_{\pi_\theta}\left[R\nabla_\theta\log\pi_\theta(A)\right]
\end{aligned}
$$

This gives us an unbiased stochastic gradient estimator and the corresponding update rule:

$$
\theta_{t+1} =
\theta_t + \alpha R_t \nabla_\theta \log \pi_{\theta_t}(A_t)
$$

Rather than evaluating the expectation over all possible actions and rewards, we can estimate the gradient using a single sampled action and its observed reward.

## Gradient Bandits

A simple policy parameterization assigns a scalar preference $H_t(a)$ to each action. These preferences are real-valued parameters and do not directly represent expected rewards.

We convert them into action probabilities using the softmax function:

$$
\pi_t(a) = \frac{\exp(H_t(a))}{\sum_b \exp(H_t(b))}
$$

For this parameterization, the gradient of the log-probability of the selected action is:

$$
\frac{\partial \log \pi_t(A_t)}{\partial H_t(a)}
=
\mathbb{1}[a=A_t] - \pi_t(a)
$$

where $\mathbb{1}[a=A_t]$ equals 1 when $a$ is the selected action and 0 otherwise.

Substituting this into the stochastic gradient update gives:

$$
H_{t+1}(a)
=
H_t(a)
+
\alpha R_t\left(\mathbb{1}[a=A_t]-\pi_t(a)\right)
$$

This update modifies the preferences of all actions, not just the selected action. When the observed reward is positive, the selected action's preference increases relative to the others; when the reward is negative, the opposite occurs.

Unlike action-value methods, gradient bandits do not explicitly estimate the expected reward of each action. Instead, they learn a probability distribution over actions directly.
